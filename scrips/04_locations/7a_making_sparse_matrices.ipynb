```python
# libraries
import pandas as pd
import numpy as np
import scipy.sparse as sparse
```

```python
# make sparse matrix function
def make_sparse_constraint_matrix(filename):
    print("Loading matrix...")
    constraint_mat = pd.read_csv(filename)
    print("Finished loading matrix")

    constraint_mat = constraint_mat.iloc[:,1:].values
    print("Constraint matrix shape:", constraint_mat.shape)

    sparse_mat = sparse.csr_matrix(- constraint_mat[1:])

    print("Saving matrix...")
    sparse_filename = filename[:-4] + "_sparse.npz"
    sparse.save_npz(sparse_filename, sparse_mat)
    print("Finished saving matrix")
    return(sparse_mat)
```
